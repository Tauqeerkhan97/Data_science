```python
import tensorflow as tf 
import numpy as np
from tensorflow import keras
```

```python
print(tf.__version__)
```

```python
# simplest sequential layer nneural network 
model = tf.keras.Sequential([keras.layers.Dense(units=1, input_shape = [1])])
```

```python
# complie the model 
model.compile(optimizer='sgd',loss='mean_squared_error')
```

```python
#make numpy arrays for data
xs = np.array([-1.0, 0.0, 2.0, 3.0, 4.0], dtype=float)
ys = np.array([-3.0, -1.0, 1.0, 3.0,7.0], dtype=float)
```

```python
#training of neural network
model.fit(xs, ys, epochs=500)
```

```python
print(model.predict(np.array([11])))
```

```python
print(model.predict(np.array([11, 10,17])))
```
