```python
# First, check what species are in your data
print("Unique species:", phool["species"].unique())
print("Species value counts:")
print(phool["species"].value_counts())

# Then create a palette with all species
species_palette = {"setosa": "seagreen", "versicolor": "gray", "virginica": "blue"}

g = sns.FacetGrid(phool, hue="species", col="petal_length", 
                  margin_titles=True, palette=species_palette)

g = g.map(plt.scatter, "sepal_length", "sepal_width", edgecolor="w").add_legend()
```

```python
# Select only numeric columns for correlation
numeric_phool = phool.select_dtypes(include=[np.number])
corr = numeric_phool.corr()

plt.figure(figsize=(10, 10))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True)
plt.title("Correlation Heatmap (Numeric Columns Only)")
plt.tight_layout()
plt.show()
```
