```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create the FacetGrid
g = sns.FacetGrid(kashti, hue="survived", col="pclass", margin_titles=True, palette={1: "seagreen", 0: "gray"})

# Map the scatter plot and add legend
g = g.map(plt.scatter, "fare", "age", edgecolor="w").add_legend()
```

```python
g = sns.FacetGrid(kashti, hue="survived", col="pclass", margin_titles=True, palette={1: "blue", 0:"red"})

g= g.map(plt.scatter, "fare", "age", edgecolor="gray").add_legend()
```

```python
g = sns.FacetGrid(kashti, hue="survived", col="who", margin_titles=True, palette={1: "blue", 0:"brown"})

g= g.map(plt.scatter, "fare", "age", edgecolor="gray").add_legend()
```

```python
sns.stripplot(data=kashti, x="sex", y="fare", hue="class")
```

```python
# heat map
corr = kashti.select_dtypes(include=['number']).corr()
plt.figure(figsize=(10,10))
sns.heatmap(corr)
plt.show()
```

```python
# heat map
corr = kashti.select_dtypes(include=['number']).corr()
plt.figure(figsize=(10,10))
sns.heatmap(corr)
plt.title("Heat Map of Kashti Data")
```
