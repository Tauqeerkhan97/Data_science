
---------|----------|------------------|----------|---------

> survived: Whether the passenger survived (0 = No, 1 = Yes)

> pclass: Passenger class (1st, 2nd, 3rd class - indicates socioeconomic status)

> sex: Gender of the passenger (male/female)

> age: Age of the passenger in years

> sibsp: Number of siblings or spouses aboard

> parch: Number of parents or children aboard

> alone: Whether the passenger was traveling alone (True/False)

---------|----------|------------------|----------|---------

> Derived/Categorical Columns:
> who: Categorization of passengers (man, woman, child)

> adult_male: Whether the passenger was an adult male (True/False)

> alive: Survival status as text (yes/no - same as survived but as object)

---------|----------|------------------|----------|---------

> Data Quality Notes:
> age has missing values (714 non-null out of 891 total)

> deck has significant missing data (only 203 non-null values)

> embarked and embark_town have 2 missing values each
