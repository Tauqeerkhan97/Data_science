```python
# Linear regression 
import pandas as pd 
import seaborn as sns

df = sns.load_dataset('titanic')
df['age'] = df['age'].fillna(df['age'].mean())
X = df[['age']]
y = df[['fare']]

# model ko create krny ky lie hum use krty hn 
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV

#creating model 
model = LinearRegression()
# Define parameter grid
param_grid = {'fit_intercept': [True, False]}

#object grid search cv (creating the model)
grid_search = GridSearchCV(model, param_grid, cv=5, scoring='r2')

#train the model 
grid_search.fit(X,y)

#y_pred = grid_search.best_estimator_.predict([[25]])


# print the best parameter 
print("best Parameters: ", grid_search.best_params_)
```

# KNN Grid Search Cv 

```python
# Linear regression 
import pandas as pd 
import seaborn as sns

df = sns.load_dataset('titanic')
df['age'] = df['age'].fillna(df['age'].mean())
X = df[['pclass', 'sex', 'age', 'sibsp', 'parch', 'fare']]
y = df[['survived']]
X = pd.get_dummies(X, columns=['sex']) # label n male ko 1 aur female ko 0 kaha dumy ki base py 
X.age.fillna(value = X['age'].mean(), inplace=True) # is ka mean nikal lena ha 


# model ko create krny ky lie hum use krty hn 
from sklearn. neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV

#creating model 
model = KNeighborsClassifier()
# Define parameter grid
param_grid = {'n_neighbors': [5,7]}

#object grid search cv (creating the model)
grid_search = GridSearchCV(model, param_grid, cv=5, scoring='f1')

#train the model 
grid_search.fit(X,y)

#y_pred = grid_search.best_estimator_.predict([[25]])


# print the best parameter 
print("best Parameters: ", grid_search.best_params_)
print("best Score: ", grid_search.best_score_)
```
